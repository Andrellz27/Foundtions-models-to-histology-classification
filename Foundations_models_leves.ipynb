```python
# -*- coding: utf-8 -*-
"""super_codigo_6 — Fine-tuning COMPLETO de 3 modelos leves em histopatologia (severidade ordinal)

Modelos (timm, pré-treinados em ImageNet-1k, todos com janela de 256 px):
  * edgenext_xs  -> edgenext_x_small.in1k                  (~2,3M params)
  * mobilevit_s  -> mobilevit_s.cvnets_in1k                (~5,6M params)
  * mobilenetv4  -> mobilenetv4_conv_small.e3600_r256_in1k (~3,8M params)

Protocolo:
  1. Split 70% treino / 10% validação / 20% teste, estratificado pela classe e feito sobre as
     IMAGENS DE ORIGEM (todas as ROIs de uma imagem ficam na mesma partição). O split é feito
     sobre os arquivos originais, ANTES de qualquer aumento de dados, e salvo em CSV.
  2. Aumento de dados RCAug aplicado ON-THE-FLY, só no treino (cada época vê versões novas das
     ROIs). Validação e teste nunca são aumentados. RCAug = mesmo Compose do notebook de
     segmentação (aplicar_aumento_de_dados): RandomRotate90, Transpose, ElasticTransform,
     GridDistortion, OpticalDistortion, flips H/V.
  3. Fine-tuning completo (todos os parâmetros treináveis, BatchNorm em modo treino),
     AdamW com LR = 1e-4, 100 épocas.
  4. O checkpoint com maior QWK de validação é usado para avaliar o teste UMA vez.
  5. Repetido em várias sementes (cada semente = split + inicialização da cabeça diferentes).

Ordem de execução no Colab: células 1 -> 8.
Se o Colab desconectar: rode tudo de novo. (modelo, semente) terminados são pulados e o treino
em andamento continua do último checkpoint salvo no Drive.

Estrutura de saída IGUAL à do super_codigo_4 (mesmos nomes de pastas, arquivos e colunas), em
resultados_histopatologia/<run_name>/:
  config_<timestamp>.json, environment.json, splits.csv, relatorio_comparativo.md
  tabelas/   splits_resumo.csv, execucoes_por_semente.csv, execucoes_desta_chamada.csv,
             resumo_media_dp.csv, resumo_ic_bootstrap.csv, resumo_ic_numerico.csv
  logs/      <tag>_seed<s>_history.csv
  predictions/ <tag>_seed<s>.csv
  graficos/  resumo_qwk_ic.png, curvas/aprendizado_<tag>.png, confusion_matrices/CM_<tag>.png
  checkpoints/  (novo: melhor modelo e estado de retomada de cada treino)
As chaves (backbone, regime, loss, lr_variant, scale, batch_cfg) seguem o mesmo formato, com
regime = "full_ft", então as tabelas podem ser concatenadas com as dos códigos anteriores.
"""
```

```python

# %% [CÉLULA 1] Configuração
import json
import math
import re
import shutil
import subprocess
import sys
import time
import warnings
import zipfile
from dataclasses import asdict, dataclass, field
from datetime import datetime
from pathlib import Path
from typing import Dict, List, Optional

import numpy as np

CLASS_NAMES = ["healthy", "mild", "moderate", "severe"]   # ordem = ordem ordinal
NUM_CLASSES = len(CLASS_NAMES)

MODELS = {
    "edgenext_xs": "edgenext_x_small.in1k",
    "mobilevit_s": "mobilevit_s.cvnets_in1k",
    "mobilenetv4": "mobilenetv4_conv_small.e3600_r256_in1k",
}


@dataclass
class Config:
    # --- caminhos ---
    dataset_zip_hint: str = "Original_ROI_images"     # parte do nome do .zip no MyDrive
    extract_dir: str = "/content/Original_ROI_images"
    local_dataset_dir: str = "/content/dataset"
    drive_results_dir: str = "/content/drive/MyDrive/resultados_histopatologia"
    run_name: str = "finetuning_completo"   # pasta do run (nome fixo => permite retomar)
    # --- experimento ---
    models: List[str] = field(default_factory=lambda: list(MODELS))
    seeds: List[int] = field(default_factory=lambda: [0, 1, 2, 3, 4])
    train_frac: float = 0.70
    val_frac: float = 0.10
    test_frac: float = 0.20
    split_by_group: bool = True        # split por imagem de origem (evita vazamento entre ROIs)
    # --- treino ---
    epochs: int = 100
    lr: float = 1e-4
    lr_schedule: str = "constante"     # "constante" (LR fixa em 1e-4) ou "cosseno" (1e-4 -> 0)
    weight_decay: float = 1e-2
    loss: str = "ce"                   # "ce", "coral" ou "corn"
    batch_size: int = 32               # ROIs por passo do otimizador (cada ROI = 2 janelas 256 px)
    grad_clip: float = 1.0
    use_amp: bool = True
    num_workers: int = 2
    patience: Optional[int] = None     # None = roda as 100 épocas inteiras (sem early stopping)
    ckpt_every: int = 5                # salva checkpoint de retomada a cada N épocas
    # --- smoke test ---
    smoke: bool = False                # True: 2 épocas, 1 semente, poucas imagens (testa o pipeline)
    smoke_groups_per_class: int = 6
    # --- estatística e saídas (iguais ao código anterior) ---
    n_boot: int = 1000
    ci_level: float = 0.95
    show_plots: bool = True


CFG = Config()
# Para um teste rápido do pipeline antes da rodada completa:
# CFG = Config(smoke=True)
if CFG.smoke:
    CFG.epochs, CFG.seeds, CFG.ckpt_every, CFG.n_boot = 2, [0], 1, 50
    CFG.run_name = f"smoke_{CFG.run_name}"
print(CFG)
```

```python

# %% [CÉLULA 2] Drive, dataset e dependências
from google.colab import drive  # noqa: E402

drive.mount("/content/drive")


def pip_install(*pkgs: str) -> None:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", *pkgs], check=True)


pip_install("timm>=1.0.9", "huggingface_hub", "scikit-learn", "pandas", "matplotlib", "tabulate",
            "albumentations")


def localizar_zip_dataset(mydrive: Path, pista: str) -> Path:
    candidatos = sorted(mydrive.rglob(f"*{pista}*.zip"))
    if len(candidatos) == 1:
        return candidatos[0]
    if not candidatos:
        raise FileNotFoundError(f"Nenhum .zip com '{pista}' em {mydrive}.")
    raise FileNotFoundError(f"Mais de um candidato: {[str(p) for p in candidatos]}. Escolha um.")


def encontrar_raiz_classes(base: Path) -> Path:
    for cand in [base] + [p for p in base.rglob("*") if p.is_dir()]:
        if all((cand / c).is_dir() for c in CLASS_NAMES):
            return cand
    raise FileNotFoundError(f"Não encontrei as pastas {CLASS_NAMES} dentro de {base}.")


extract_dir = Path(CFG.extract_dir)
if not extract_dir.exists():
    zp = localizar_zip_dataset(Path("/content/drive/MyDrive"), CFG.dataset_zip_hint)
    print(f"[dataset] extraindo {zp}")
    with zipfile.ZipFile(zp) as z:
        z.extractall(extract_dir)
DATA_ROOT = encontrar_raiz_classes(extract_dir)
print("Raiz do dataset:", DATA_ROOT)
```

```python

# %% [CÉLULA 3] Imports, dataset e split 70/10/20 (antes de qualquer aumento de dados)
import random  # noqa: E402

import pandas as pd  # noqa: E402
import timm  # noqa: E402
import torch  # noqa: E402
import torch.nn as nn  # noqa: E402
import torch.nn.functional as F  # noqa: E402
import torchvision.transforms as T  # noqa: E402
from PIL import Image  # noqa: E402
from sklearn.exceptions import UndefinedMetricWarning  # noqa: E402
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, cohen_kappa_score,  # noqa: E402
                             confusion_matrix, f1_score, roc_auc_score)
from sklearn.model_selection import train_test_split  # noqa: E402
from torch.utils.data import DataLoader, Dataset  # noqa: E402

warnings.filterwarnings("ignore", category=UndefinedMetricWarning)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
assert DEVICE.type == "cuda", "Ative uma GPU no Colab (Ambiente de execução > Alterar tipo)."
# bf16 só em GPUs com suporte NATIVO (Ampere+, capability >= 8.0). No T4 (7.5) as versões novas do
# torch dizem que bf16 é "suportado" (por emulação), mas o cuDNN não tem kernels de convolução em
# bf16 para essa GPU -> "GET was unable to find an engine to execute this computation".
GPU_CC = torch.cuda.get_device_capability(0)
AMP_DTYPE = torch.bfloat16 if GPU_CC >= (8, 0) else torch.float16
CUDNN_DETERMINISTIC = True     # pode ser desligado automaticamente por check_cudnn() (ver célula 7)
print(f"[gpu] {torch.cuda.get_device_name(0)} (capability {GPU_CC[0]}.{GPU_CC[1]}) | AMP em {AMP_DTYPE}")
IMG_EXTS = {".png", ".jpg", ".jpeg", ".tif", ".tiff", ".bmp"}
ROI_RE = re.compile(r"(.*)-roi\d+$")


def list_images(d: Path) -> List[Path]:
    return sorted(p for p in d.rglob("*") if p.suffix.lower() in IMG_EXTS)


def copy_to_local(src: Path, dst: Path) -> Path:
    """Cópia para o disco local do Colab (leitura muito mais rápida que no Drive)."""
    n_src = sum(len(list_images(src / c)) for c in CLASS_NAMES)
    n_dst = sum(len(list_images(dst / c)) for c in CLASS_NAMES if (dst / c).is_dir())
    if n_src != n_dst:
        for c in CLASS_NAMES:
            shutil.copytree(src / c, dst / c, dirs_exist_ok=True)
    return dst


def scan_dataset(root: Path):
    paths, labels = [], []
    for k, c in enumerate(CLASS_NAMES):
        files = list_images(root / c)
        assert files, f"Nenhuma imagem em {root / c}"
        paths += [str(p) for p in files]
        labels += [k] * len(files)
    return paths, np.array(labels, dtype=np.int64)


def group_id(path: str) -> str:
    """'healthy/image032-roi1' -> 'healthy/image032' (classe + imagem de origem)."""
    p = Path(path)
    m = ROI_RE.match(p.stem)
    return f"{p.parent.name}/{m.group(1) if m else p.stem}"


def make_split(labels: np.ndarray, groups: np.ndarray, seed: int, cfg: Config) -> Dict[str, np.ndarray]:
    """70/10/20 estratificado. Com split_by_group as unidades sorteadas são as imagens de origem."""
    fr = (cfg.train_frac, cfg.val_frac, cfg.test_frac)
    assert abs(sum(fr) - 1) < 1e-6, fr
    if cfg.split_by_group:
        units, first = np.unique(groups, return_index=True)
        unit_y = labels[first]
    else:
        units, unit_y = np.arange(len(labels)), labels
    u = np.arange(len(units))
    u_tv, u_te = train_test_split(u, test_size=cfg.test_frac, stratify=unit_y, random_state=seed)
    u_tr, u_va = train_test_split(u_tv, test_size=cfg.val_frac / (cfg.train_frac + cfg.val_frac),
                                  stratify=unit_y[u_tv], random_state=seed)

    def to_idx(uu):
        return np.flatnonzero(np.isin(groups, units[uu])) if cfg.split_by_group else np.sort(units[uu])

    sp = {"train": to_idx(u_tr), "val": to_idx(u_va), "test": to_idx(u_te)}
    allidx = np.concatenate(list(sp.values()))
    assert len(allidx) == len(labels) == len(np.unique(allidx)), "split não particiona o dataset"
    if cfg.split_by_group:
        g = {k: set(groups[v]) for k, v in sp.items()}
        assert not (g["train"] & g["val"] or g["train"] & g["test"] or g["val"] & g["test"]), "vazamento"
    return sp


LOCAL_ROOT = copy_to_local(DATA_ROOT, Path(CFG.local_dataset_dir))
PATHS, LABELS = scan_dataset(LOCAL_ROOT)
GROUPS = np.array([group_id(p) for p in PATHS])
if CFG.smoke:
    rng = np.random.default_rng(0)
    keep = []
    for c in range(NUM_CLASSES):
        g = np.unique(GROUPS[LABELS == c])
        rng.shuffle(g)
        keep += list(g[:CFG.smoke_groups_per_class])
    idx = np.flatnonzero(np.isin(GROUPS, keep))
    PATHS, LABELS, GROUPS = [PATHS[i] for i in idx], LABELS[idx], GROUPS[idx]
ROI_WH = Image.open(PATHS[0]).size
print(f"[dataset] {len(PATHS)} ROIs de {len(set(GROUPS))} imagens | por classe: "
      f"{dict(zip(CLASS_NAMES, np.bincount(LABELS).tolist()))} | ROI {ROI_WH[0]}x{ROI_WH[1]}")

# --- Chaves e nomes de arquivo no MESMO formato do super_codigo_4 ---
REGIME = "full_ft"
KEY_COLS = ["backbone", "regime", "loss", "lr_variant", "scale", "batch_cfg"]
PT_NAMES = {"backbone": "Modelo", "regime": "Regime", "loss": "Loss", "lr_variant": "LR",
            "scale": "Escala", "batch_cfg": "Config. batch"}


def _slug(s: str) -> str:
    return "".join(ch if ch.isalnum() or ch in "-_.|" else "_" for ch in s).replace("|", "__")


def key_tag(key) -> str:
    return _slug("|".join(str(k) for k in key))


def norm_key(key) -> tuple:
    bb, reg, loss, lrv, sc, bc = key
    return (str(bb), str(reg), str(loss), str(lrv), round(float(sc), 6), str(bc))


def scenario_key(model_key: str, cfg: Config) -> tuple:
    return norm_key((model_key, REGIME, cfg.loss, f"{cfg.lr:g}", 1.0,
                     f"bs{cfg.batch_size}_eff{cfg.batch_size}"))


def split_summary(splits, labels, groups) -> pd.DataFrame:
    rows = []
    for seed, sp in splits.items():
        r = {"semente": seed}
        for part in ("train", "val", "test"):
            idx = sp[part]
            r[f"{part}_rois"] = len(idx)
            r[f"{part}_%"] = round(100 * len(idx) / len(labels), 1)
            r[f"{part}_imagens"] = len(set(groups[idx]))
            r[f"{part}_por_classe"] = np.bincount(labels[idx], minlength=NUM_CLASSES).tolist()
        rows.append(r)
    return pd.DataFrame(rows)


def collect_env() -> Dict:
    import importlib.metadata as imd
    pkgs = {}
    for name in ("timm", "torch", "torchvision", "numpy", "scikit-learn", "pandas", "huggingface-hub"):
        try:
            pkgs[name] = imd.version(name)
        except imd.PackageNotFoundError:
            pass
    hf_sha = {}
    try:
        from huggingface_hub import HfApi
        api = HfApi()
        for k, v in MODELS.items():
            try:
                hf_sha[k] = api.model_info(f"timm/{v}").sha
            except Exception as e:
                hf_sha[k] = f"indisponível ({type(e).__name__})"
    except ImportError:
        pass
    return {"python": sys.version.split()[0], "packages": pkgs,
            "gpu": torch.cuda.get_device_name(0),
            "vram_gb": round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2),
            "cuda": torch.version.cuda, "amp_dtype": str(AMP_DTYPE), "cudnn_deterministic": CUDNN_DETERMINISTIC,
            "hf_sha": hf_sha, "aumento_de_dados": "RCAug (on-the-fly, só treino)"}


ENV_INFO = collect_env()
TS = f"{datetime.now():%Y%m%d_%H%M%S}"
OUT = Path(CFG.drive_results_dir) / CFG.run_name
for sub in ("tabelas", "graficos/confusion_matrices", "graficos/curvas", "logs", "predictions", "checkpoints"):
    (OUT / sub).mkdir(parents=True, exist_ok=True)
(OUT / f"config_{TS}.json").write_text(json.dumps(asdict(CFG), indent=2, default=str))
(OUT / "environment.json").write_text(json.dumps(ENV_INFO, indent=2))
TABLES_DIR, LOGS_DIR, PRED_DIR = OUT / "tabelas", OUT / "logs", OUT / "predictions"
PLOTS_DIR, CKPT_DIR = OUT / "graficos", OUT / "checkpoints"

SPLITS = {s: make_split(LABELS, GROUPS, s, CFG) for s in CFG.seeds}
split_csv = OUT / "splits.csv"
new_split = pd.DataFrame([{"seed": s, "part": part, "idx": i, "path": PATHS[i], "group": GROUPS[i]}
                          for s, sp in SPLITS.items() for part, ix in sp.items() for i in ix])
if split_csv.exists():
    new_split = pd.concat([pd.read_csv(split_csv), new_split]).drop_duplicates(["seed", "idx"])
new_split.to_csv(split_csv, index=False)
resumo_split = split_summary(SPLITS, LABELS, GROUPS)
resumo_split.to_csv(TABLES_DIR / "splits_resumo.csv", index=False)
print(f"Split {CFG.train_frac:.0%}/{CFG.val_frac:.0%}/{CFG.test_frac:.0%} por semente "
      f"({'por imagem de origem' if CFG.split_by_group else 'por ROI'}):")
print(resumo_split.to_string(index=False))
```

```python

# %% [CÉLULA 4] Aumento de dados RCAug (em tempo real, só no treino)
import albumentations as A  # noqa: E402

# Mesmo aumento do notebook de segmentação (aplicar_aumento_de_dados), sem a máscara:
# na classificação só a imagem é transformada.
def build_rcaug():
    transformacoes = A.Compose([
        A.RandomRotate90(p=0.5),
        A.Transpose(p=0.5),
        A.ElasticTransform(p=0.3, alpha=120, sigma=120 * 0.05),
        A.GridDistortion(p=0.3),
        A.OpticalDistortion(p=0.3, distort_limit=2),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
    ])

    def aplicar(img: Image.Image) -> np.ndarray:
        return transformacoes(image=np.asarray(img))["image"]

    return aplicar


AUG_NAME = "RCAug"


def _to_pil(out) -> Image.Image:
    if isinstance(out, Image.Image):
        return out.convert("RGB")
    arr = np.asarray(out)
    if arr.dtype != np.uint8:   # aceita float em [0, 1]
        arr = (np.clip(arr, 0, 1) * 255).round().astype(np.uint8) if arr.max() <= 1.0 else arr.astype(np.uint8)
    return Image.fromarray(arr).convert("RGB")


# Testa o RCAug numa ROI real antes de gastar horas de GPU (várias vezes, pois é aleatório)
RCAUG = build_rcaug()
_formatos = {_to_pil(RCAUG(Image.open(PATHS[0]).convert("RGB"))).size for _ in range(20)}
print(f"[RCAug] ok: ROI {ROI_WH} -> formatos gerados em 20 sorteios: {sorted(_formatos)}")
```

```python

# %% [CÉLULA 5] Multi-crop, Dataset e modelo
def roi_views(img: Image.Image, size: int) -> List[Image.Image]:
    """Redimensiona o LADO MENOR da ROI para `size` e extrai janelas size x size cobrindo a ROI
    inteira (450x250 -> 2 janelas de 256). Como o RCAug pode girar a ROI em 90 graus
    (RandomRotate90/Transpose: 450x250 -> 250x450), o corte segue o lado maior, seja ele largura
    ou altura: a ROI girada também vira 2 janelas, na MESMA escala da validação/teste."""
    w, h = img.size
    sc = size / min(w, h)
    img = img.resize((max(size, round(w * sc)), max(size, round(h * sc))), Image.BICUBIC)
    arr = np.asarray(img)
    H, W = arr.shape[:2]
    ys = np.linspace(0, H - size, num=max(1, math.ceil(H / size))).round().astype(int)
    xs = np.linspace(0, W - size, num=max(1, math.ceil(W / size))).round().astype(int)
    return [Image.fromarray(np.ascontiguousarray(arr[y:y + size, x:x + size])) for y in ys for x in xs]


class ROIDataset(Dataset):
    def __init__(self, paths, labels, size: int, normalize, aug=None):
        self.paths, self.labels, self.size = paths, labels, size
        self.normalize, self.aug = normalize, aug

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, i):
        img = Image.open(self.paths[i]).convert("RGB")
        if self.aug is not None:
            img = _to_pil(self.aug(img))          # aumento em tempo real (só treino)
        views = roi_views(img, self.size)
        return torch.stack([self.normalize(v) for v in views]), int(self.labels[i]), i


def collate_views(batch):
    xs, ys, idx = zip(*batch)
    return (torch.cat(xs), torch.tensor([x.shape[0] for x in xs]),
            torch.tensor(ys, dtype=torch.long), torch.tensor(idx, dtype=torch.long))


def _seed_worker(_):
    s = torch.initial_seed() % 2**32
    np.random.seed(s)
    random.seed(s)


def make_loader(ds, batch_size, shuffle, seed, cfg):
    g = torch.Generator()
    g.manual_seed(seed)
    return DataLoader(ds, batch_size=batch_size, shuffle=shuffle, num_workers=cfg.num_workers,
                      pin_memory=True, generator=g, worker_init_fn=_seed_worker,
                      collate_fn=collate_views, persistent_workers=cfg.num_workers > 0,
                      drop_last=False)


def set_seed(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = CUDNN_DETERMINISTIC
    torch.backends.cudnn.benchmark = False


def model_input_spec(backbone: nn.Module):
    """Tamanho da janela e normalização oficiais do checkpoint (pretrained_cfg do timm).
    MobileViT: mean=0/std=1; MobileNetV4-S r256: 0,5/0,5; EdgeNeXt: ImageNet."""
    cfg = backbone.pretrained_cfg
    return int(cfg["input_size"][-1]), T.Compose([T.ToTensor(), T.Normalize(cfg["mean"], cfg["std"])])


class Classifier(nn.Module):
    """Backbone timm completo (treinável) + cabeça linear sobre a média das janelas da ROI."""

    def __init__(self, timm_name: str, loss: str):
        super().__init__()
        self.backbone = timm.create_model(timm_name, pretrained=True, num_classes=0)
        out = NUM_CLASSES if loss == "ce" else NUM_CLASSES - 1
        # Dimensão REAL do vetor que o backbone devolve com num_classes=0 (pre-logits).
        # Não usar backbone.num_features: no MobileNetV4 ele vale 960 (saída do último estágio),
        # mas o vetor final passa pelo conv_head e tem 1280 dimensões (head_hidden_size).
        self.feat_dim = self._feature_dim()
        self.head = nn.Linear(self.feat_dim, out)

    @torch.no_grad()
    def _feature_dim(self) -> int:
        size = int(self.backbone.pretrained_cfg["input_size"][-1])
        was_training = self.backbone.training
        self.backbone.eval()
        d = int(self.backbone(torch.zeros(1, 3, size, size)).shape[-1])
        self.backbone.train(was_training)
        attr = getattr(self.backbone, "head_hidden_size", None) or self.backbone.num_features
        if attr != d:
            print(f"   [info] {type(self.backbone).__name__}: num_features={self.backbone.num_features}, "
                  f"saída real={d} -> cabeça usa {d}")
        return d

    def forward(self, x, counts):
        z = self.backbone(x)                                             # (janelas) x D
        z = torch.stack([c.mean(0) for c in torch.split(z, counts.tolist())])
        return self.head(z)

```

```python

# %% [CÉLULA 6] Perdas, decodificação e métricas
def coral_loss(logits, y):
    levels = (y.unsqueeze(1) > torch.arange(NUM_CLASSES - 1, device=y.device)).float()
    ls = F.logsigmoid(logits)
    return -(ls * levels + (ls - logits) * (1 - levels)).sum(1).mean()


def corn_loss(logits, y):
    total, n = logits.new_zeros(()), 0
    for k in range(NUM_CLASSES - 1):
        m = y >= k
        if m.any():
            total = total + F.binary_cross_entropy_with_logits(logits[m, k], (y[m] > k).float(), reduction="sum")
            n += int(m.sum())
    return total / max(n, 1)


def make_criterion(name):
    return {"ce": nn.CrossEntropyLoss(), "coral": coral_loss, "corn": corn_loss}[name]


def probs_preds(logits: np.ndarray, loss: str):
    logits = np.asarray(logits, dtype=np.float64)
    if loss == "ce":
        e = np.exp(logits - logits.max(1, keepdims=True))
        P = e / e.sum(1, keepdims=True)
        return P, P.argmax(1)
    S = 1 / (1 + np.exp(-logits))
    cum = S if loss == "coral" else np.cumprod(S, axis=1)               # P(y > k)
    n = len(cum)
    P = np.clip(np.hstack([np.ones((n, 1)), cum]) - np.hstack([cum, np.zeros((n, 1))]), 0, None)
    return P / np.clip(P.sum(1, keepdims=True), 1e-12, None), (cum > 0.5).sum(1)


METRIC_KEYS = ("qwk", "f1_macro", "bacc", "acc", "auroc", "mod")


def compute_metrics(y, P, d) -> Dict[str, float]:
    labels = list(range(NUM_CLASSES))
    try:
        auroc = roc_auc_score(y, P, multi_class="ovr", average="macro", labels=labels)
    except ValueError:
        auroc = float("nan")
    return {"qwk": cohen_kappa_score(y, d, weights="quadratic", labels=labels),
            "f1_macro": f1_score(y, d, average="macro", labels=labels, zero_division=0),
            "bacc": balanced_accuracy_score(y, d), "acc": accuracy_score(y, d), "auroc": auroc,
            "mod": float(np.abs(np.asarray(y) - np.asarray(d)).mean())}


def _safe(v):
    return v if np.isfinite(v) else -2.0

```

```python

# %% [CÉLULA 7] Treino (fine-tuning completo) com retomada
@torch.no_grad()
def predict(model, loader, loss_name, criterion, cfg):
    model.eval()
    ys, logits, tot = [], [], 0.0
    for x, counts, y, _ in loader:
        with torch.autocast("cuda", dtype=AMP_DTYPE, enabled=cfg.use_amp):
            lg = model(x.to(DEVICE, non_blocking=True), counts).float()
        tot += float(criterion(lg, y.to(DEVICE))) * len(y)
        ys.append(y.numpy())
        logits.append(lg.cpu().numpy())
    y = np.concatenate(ys)
    P, d = probs_preds(np.concatenate(logits), loss_name)
    return y, P, d, tot / len(y)


def _rng_state():
    return {"torch": torch.get_rng_state(), "cuda": torch.cuda.get_rng_state_all(),
            "numpy": np.random.get_state(), "python": random.getstate()}


def _set_rng_state(s):
    torch.set_rng_state(s["torch"])
    torch.cuda.set_rng_state_all(s["cuda"])
    np.random.set_state(s["numpy"])
    random.setstate(s["python"])


_CUDNN_OK = set()


def check_cudnn(model: nn.Module, size: int, model_key: str, cfg: Config) -> None:
    """Forward + backward de teste (1 ROI = 2 janelas) antes do treino. Se o cuDNN não tiver kernel
    determinístico para alguma convolução (comum em convoluções depthwise com AMP), desliga o modo
    determinístico e tenta de novo. Roda em eval (não altera as estatísticas do BatchNorm) e não
    dá passo no otimizador, então os pesos não mudam."""
    global CUDNN_DETERMINISTIC
    if model_key in _CUDNN_OK:
        return
    x = torch.full((2, 3, size, size), 0.5, device=DEVICE)
    counts, y = torch.tensor([2]), torch.zeros(1, dtype=torch.long, device=DEVICE)
    crit = make_criterion(cfg.loss)
    for tentativa in range(2):
        try:
            model.eval()
            with torch.autocast("cuda", dtype=AMP_DTYPE, enabled=cfg.use_amp):
                out = model(x, counts)
            crit(out.float(), y).backward()
            model.zero_grad(set_to_none=True)
            _CUDNN_OK.add(model_key)
            return
        except RuntimeError as e:
            model.zero_grad(set_to_none=True)
            if "find an engine" in str(e) and tentativa == 0 and CUDNN_DETERMINISTIC:
                print(f"   [aviso] {model_key}: cuDNN sem kernel determinístico -> "
                      "cudnn.deterministic=False (resultados podem variar levemente entre execuções)")
                CUDNN_DETERMINISTIC = False
                torch.backends.cudnn.deterministic = False
                ENV_INFO["cudnn_deterministic"] = False
                (OUT / "environment.json").write_text(json.dumps(ENV_INFO, indent=2))
                continue
            raise


def train_one(model_key: str, seed: int, cfg: Config) -> Dict:
    timm_name = MODELS[model_key]
    key = scenario_key(model_key, cfg)
    tag = f"{key_tag(key)}_seed{seed}"
    ck_resume = CKPT_DIR / f"{tag}_retomada.pt"
    ck_best = CKPT_DIR / f"{tag}_melhor.pt"
    sp = SPLITS[seed]

    set_seed(seed)
    model = Classifier(timm_name, cfg.loss).to(DEVICE)
    size, normalize = model_input_spec(model.backbone)
    check_cudnn(model, size, model_key, cfg)
    n_par = sum(p.numel() for p in model.parameters())
    optimizer = torch.optim.AdamW(model.parameters(), lr=cfg.lr, weight_decay=cfg.weight_decay)
    criterion = make_criterion(cfg.loss)
    scaler = torch.amp.GradScaler("cuda", enabled=cfg.use_amp and AMP_DTYPE == torch.float16)

    def mk(part, aug):
        ix = sp[part]
        return ROIDataset([PATHS[i] for i in ix], LABELS[ix], size, normalize, aug)

    train_loader = make_loader(mk("train", RCAUG), cfg.batch_size, True, seed, cfg)
    val_loader = make_loader(mk("val", None), cfg.batch_size * 2, False, seed, cfg)
    test_loader = make_loader(mk("test", None), cfg.batch_size * 2, False, seed, cfg)

    total_steps = cfg.epochs * len(train_loader)
    sched = torch.optim.lr_scheduler.LambdaLR(
        optimizer, (lambda s: 1.0) if cfg.lr_schedule == "constante" else
        (lambda s: 0.5 * (1 + math.cos(math.pi * min(s, total_steps) / total_steps))))

    start_ep, best_q, best_vloss, best_ep, bad, history = 0, -np.inf, np.inf, 0, 0, []
    if ck_resume.exists():
        st = torch.load(ck_resume, map_location="cpu", weights_only=False)
        model.load_state_dict(st["model"])
        optimizer.load_state_dict(st["optimizer"])
        sched.load_state_dict(st["sched"])
        scaler.load_state_dict(st["scaler"])
        start_ep, best_q, best_vloss, best_ep, bad, history = (st["epoch"], st["best_q"], st["best_vloss"],
                                                               st["best_ep"], st["bad"], st["history"])
        _set_rng_state(st["rng"])
        print(f"   [retomada] {tag}: continuando da época {start_ep + 1}")

    print(f"   {model_key} ({timm_name}) | {n_par / 1e6:.2f}M params treináveis | janela {size}px | "
          f"embedding {model.feat_dim}-d | "
          f"treino={len(sp['train'])} val={len(sp['val'])} teste={len(sp['test'])} ROIs")
    torch.cuda.reset_peak_memory_stats()
    t_start = time.time() - (history[-1]["cum_time_s"] if history else 0.0)
    for ep in range(start_ep, cfg.epochs):
        te = time.time()
        model.train()                              # fine-tuning completo: BN e dropout em modo treino
        run, n = 0.0, 0
        for x, counts, y, _ in train_loader:
            x, y = x.to(DEVICE, non_blocking=True), y.to(DEVICE, non_blocking=True)
            optimizer.zero_grad(set_to_none=True)
            with torch.autocast("cuda", dtype=AMP_DTYPE, enabled=cfg.use_amp):
                logits = model(x, counts)
            loss = criterion(logits.float(), y)
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), cfg.grad_clip)
            scaler.step(optimizer)
            scaler.update()
            sched.step()
            run += float(loss) * len(y)
            n += len(y)

        yv, Pv, dv, vloss = predict(model, val_loader, cfg.loss, criterion, cfg)
        vm = compute_metrics(yv, Pv, dv)
        history.append({"epoch": ep + 1, "train_loss": run / n, "val_loss": vloss,
                         **{f"val_{k}": v for k, v in vm.items()},
                         "lr_head": optimizer.param_groups[0]["lr"], "epoch_time_s": time.time() - te,
                         "cum_time_s": time.time() - t_start})
        q = _safe(vm["qwk"])
        improved = q > best_q or (q == best_q and vloss < best_vloss)   # empate no QWK -> menor val_loss
        if improved:
            best_q, best_vloss, best_ep, bad = q, vloss, ep + 1, 0
            torch.save({"model": model.state_dict(), "epoch": ep + 1, "val_qwk": q}, ck_best)
        else:
            bad += 1
        print(f"      ep {ep + 1:03d} loss={run / n:.4f} val_loss={vloss:.4f} val_qwk={vm['qwk']:.4f} "
              f"{'*' if improved else ' '} ({time.time() - te:.1f}s)")

        last = ep + 1 == cfg.epochs
        stop = cfg.patience is not None and bad >= cfg.patience
        if (ep + 1) % cfg.ckpt_every == 0 or last or stop:
            torch.save({"model": model.state_dict(), "optimizer": optimizer.state_dict(),
                        "sched": sched.state_dict(), "scaler": scaler.state_dict(), "epoch": ep + 1,
                        "best_q": best_q, "best_vloss": best_vloss, "best_ep": best_ep, "bad": bad,
                        "history": history, "rng": _rng_state()}, ck_resume)
        if stop:
            print(f"      early stopping na época {ep + 1} (melhor: {best_ep})")
            break

    # avaliação final: melhor época pela validação, teste avaliado uma única vez
    model.load_state_dict(torch.load(ck_best, map_location="cpu", weights_only=False)["model"])
    yv, Pv, dv, _ = predict(model, val_loader, cfg.loss, criterion, cfg)
    yt, Pt, dt, _ = predict(model, test_loader, cfg.loss, criterion, cfg)
    hist = pd.DataFrame(history)
    hist.assign(**dict(zip(KEY_COLS, key)), seed=seed, batch_size=cfg.batch_size, accum_steps=1).to_csv(
        LOGS_DIR / f"{tag}_history.csv", index=False)
    ix = sp["test"]
    pd.DataFrame({"idx": ix, "path": [PATHS[i] for i in ix], "group": GROUPS[ix], "y_true": yt, "y_pred": dt,
                  **{f"p_{c}": Pt[:, k] for k, c in enumerate(CLASS_NAMES)}}
                 ).to_csv(PRED_DIR / f"{tag}.csv", index=False)
    vm, tm = compute_metrics(yv, Pv, dv), compute_metrics(yt, Pt, dt)
    extra = {"batch_size": cfg.batch_size, "accum_steps": 1, "effective_batch": cfg.batch_size,
             "batch_role": "treino", "best_epoch": best_ep, "epochs_run": len(hist),
             "train_time_s": float(hist.cum_time_s.iloc[-1]),
             "time_to_best_s": float(hist.loc[hist.epoch == best_ep, "cum_time_s"].iloc[0]),
             "time_per_epoch_s": float(hist.epoch_time_s.mean()),
             "peak_vram_gb": torch.cuda.max_memory_allocated() / 1024**3, "oom_retries": 0,
             "C": None, "lbfgs_iter": None, "timm_id": timm_name, "params_M": n_par / 1e6}
    rec = _record(key, seed, extra, vm, tm)
    ck_resume.unlink(missing_ok=True)              # treino concluído: libera espaço no Drive
    del model, optimizer, scaler
    torch.cuda.empty_cache()
    return rec


def _record(key, seed, extra: Dict, vm: Dict, tm: Dict) -> Dict:
    """Mesma linha de execucoes_por_semente.csv do super_codigo_4."""
    base = {c: None for c in ("batch_size", "accum_steps", "effective_batch", "batch_role", "best_epoch",
                              "epochs_run", "train_time_s", "time_to_best_s", "time_per_epoch_s",
                              "peak_vram_gb", "oom_retries", "C", "lbfgs_iter")}
    base.update(extra)
    return {**dict(zip(KEY_COLS, key)), "seed": seed, **base,
            **{f"test_{k}": tm[k] for k in METRIC_KEYS}, **{f"val_{k}": vm[k] for k in METRIC_KEYS},
            "timestamp": f"{datetime.now():%Y-%m-%d %H:%M:%S}"}


def run_all(cfg: Config):
    runs_csv = TABLES_DIR / "execucoes_por_semente.csv"
    records = pd.read_csv(runs_csv).to_dict("records") if runs_csv.exists() else []
    done = {norm_key(tuple(r[c] for c in KEY_COLS)) + (int(r["seed"]),) for r in records}
    plan = [(scenario_key(m, cfg), m, s) for m in cfg.models for s in cfg.seeds]
    todo = [p for p in plan if p[0] + (p[2],) not in done]
    touched = list(dict.fromkeys(k for k, _, _ in plan))
    if done:
        print(f"[retomada] {len(done)} execução(ões) já salvas em {OUT.name} serão puladas")
    print(f"[plano] {len(touched)} cenário(s), {len(plan)} execução(ões), {len(todo)} pendente(s)")
    durations = []
    for n_done, (key, m, s) in enumerate(todo, 1):
        print(f"\n  >> {' | '.join(map(str, key))} | semente {s}")
        t0 = time.time()
        rec = train_one(m, s, cfg)
        records.append(rec)
        pd.DataFrame(records).to_csv(runs_csv, index=False)     # salva a cada execução
        durations.append(time.time() - t0)
        eta = np.mean(durations[-10:]) * (len(todo) - n_done)
        print(f"     teste: QWK={rec['test_qwk']:.4f} F1={rec['test_f1_macro']:.4f} AUROC={rec['test_auroc']:.4f} | "
              f"val QWK={rec['val_qwk']:.4f} | melhor época {rec['best_epoch']} | {n_done}/{len(todo)} "
              f"| ETA ~{eta / 60:.0f} min")
    if not records:
        print("[aviso] nenhum resultado para consolidar")
        return None
    return consolidate(cfg, pd.DataFrame(records), touched)

```

```python

# %% [CÉLULA 8] Consolidação (tabelas, IC por bootstrap, gráficos e relatório) — igual ao super_codigo_4
import matplotlib.pyplot as plt  # noqa: E402
from scipy.stats import rankdata  # noqa: E402

try:
    from IPython.display import display  # noqa: E402
except ImportError:  # fora do notebook
    def display(obj):
        print(obj.to_string() if hasattr(obj, "to_string") else obj)


def save_table(df: pd.DataFrame, path: Path, title: str, show: bool = True) -> None:
    """Exporta imediatamente para o Drive e exibe."""
    path.parent.mkdir(parents=True, exist_ok=True)
    df.to_csv(path, index=False)
    if show:
        print(f"\n===== {title} =====  ->  {path}")
        display(df)


def show_and_save(fig, path: Path, cfg: Config) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(path, dpi=150, bbox_inches="tight")
    if cfg.show_plots:
        plt.show()
    plt.close(fig)


def fast_metrics(y_true, probs, y_pred) -> Dict[str, float]:
    """MESMAS métricas de compute_metrics, direto da matriz de confusão (usado no bootstrap)."""
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    P = np.asarray(probs, dtype=np.float64)
    P = P / P.sum(1, keepdims=True)
    K, n = NUM_CLASSES, len(y_true)
    cm = np.bincount(y_true * K + y_pred, minlength=K * K).reshape(K, K).astype(np.float64)
    tp, true_n, pred_n = np.diag(cm), cm.sum(1), cm.sum(0)
    w = (np.arange(K)[:, None] - np.arange(K)[None, :]) ** 2
    den = (w * np.outer(true_n, pred_n) / n).sum()
    qwk = 1 - (w * cm).sum() / den if den > 0 else float("nan")
    f1d = 2 * tp + (pred_n - tp) + (true_n - tp)
    f1 = np.where(f1d > 0, 2 * tp / np.where(f1d > 0, f1d, 1), 0.0).mean()
    present = true_n > 0
    bacc = (tp[present] / true_n[present]).mean()
    aucs = []
    for c in range(K):
        pos = y_true == c
        n_pos = int(pos.sum())
        if n_pos == 0 or n_pos == n:
            aucs.append(float("nan"))
            continue
        r = rankdata(P[:, c])
        aucs.append((r[pos].sum() - n_pos * (n_pos + 1) / 2) / (n_pos * (n - n_pos)))
    return {"qwk": float(qwk), "f1_macro": float(f1), "bacc": float(bacc),
            "acc": float(tp.sum() / n), "auroc": float(np.mean(aucs)),
            "mod": float(np.abs(y_true - y_pred).mean())}


# --- Estatística: média entre sementes e IC por bootstrap de imagens de origem ---
def seed_mean_metrics(test_by_seed, metric_fn=compute_metrics) -> Dict[str, float]:
    per = [metric_fn(y, P, d) for (_, y, P, d) in test_by_seed.values()]
    return {k: float(np.nanmean([m[k] for m in per])) for k in METRIC_KEYS}


def holdout_bootstrap_ci(test_by_seed, groups: np.ndarray, n_boot: int, level: float, seed: int = 0):
    """IC percentil da métrica MÉDIA entre sementes, reamostrando IMAGENS DE ORIGEM do teste de
    cada semente, estratificado pela classe."""
    rng = np.random.default_rng(seed)
    prepared = []
    for idx, y, P, d in test_by_seed.values():
        uniq, inv = np.unique(groups[idx], return_inverse=True)
        members = [np.flatnonzero(inv == g) for g in range(len(uniq))]
        by_class = {}
        for g, m in enumerate(members):
            by_class.setdefault(int(y[m[0]]), []).append(g)
        prepared.append((y, P, d, members, list(by_class.values())))
    point = seed_mean_metrics(test_by_seed)
    boots = {k: np.empty(n_boot) for k in METRIC_KEYS}
    for b in range(n_boot):
        per = []
        for y, P, d, members, strata in prepared:
            pick = [g for st in strata for g in rng.choice(st, size=len(st), replace=True)]
            sel = np.concatenate([members[g] for g in pick])
            per.append(fast_metrics(y[sel], P[sel], d[sel]))
        for k in METRIC_KEYS:
            vals = [m[k] for m in per]
            boots[k][b] = np.nanmean(vals) if np.isfinite(vals).any() else np.nan
    a = (1 - level) / 2 * 100
    return {k: (point[k], float(np.nanpercentile(boots[k], a)), float(np.nanpercentile(boots[k], 100 - a)))
            for k in METRIC_KEYS}


# --- Gráficos ---
def plot_learning_curves(key, histories, best_epochs, path: Path, cfg: Config) -> None:
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    for i, (seed, h) in enumerate(sorted(histories.items())):
        col = plt.cm.tab10(i % 10)
        axes[0].plot(h.epoch, h.train_loss, "-", color=col, label=f"treino s{seed}")
        axes[0].plot(h.epoch, h.val_loss, "--", color=col, label=f"val s{seed}")
        axes[1].plot(h.epoch, h.val_qwk, "-", color=col, label=f"s{seed}")
        axes[2].plot(h.epoch, h.val_f1_macro, "-", color=col, label=f"s{seed}")
        be = best_epochs.get(seed)
        if be is not None and (h.epoch == be).any():
            axes[1].plot(be, h.loc[h.epoch == be, "val_qwk"].iloc[0], "o", color=col, ms=7)
    for ax, t in zip(axes, ["Loss (— treino, -- validação)", "QWK validação (● = época escolhida)",
                            "F1 macro validação"]):
        ax.set_title(t, fontsize=9)
        ax.set_xlabel("Época")
        ax.grid(alpha=0.3)
    axes[0].legend(fontsize=6, ncol=2)
    fig.suptitle(" | ".join(map(str, key)), fontsize=10)
    fig.tight_layout()
    show_and_save(fig, path, cfg)


def plot_confusion(key, cm: np.ndarray, n_seeds: int, path: Path, cfg: Config) -> None:
    norm = cm / cm.sum(1, keepdims=True).clip(min=1)
    fig, ax = plt.subplots(figsize=(4.8, 4.2))
    ax.imshow(norm, vmin=0, vmax=1, cmap="Blues")
    ax.set_xticks(range(NUM_CLASSES), CLASS_NAMES, rotation=45)
    ax.set_yticks(range(NUM_CLASSES), CLASS_NAMES)
    for i in range(NUM_CLASSES):
        for j in range(NUM_CLASSES):
            ax.text(j, i, f"{norm[i, j]:.2f}\n({cm[i, j]})", ha="center", va="center", fontsize=7,
                    color="white" if norm[i, j] > 0.6 else "black")
    ax.set_xlabel("Predito")
    ax.set_ylabel("Real")
    ax.set_title(" | ".join(map(str, key)) + f"\nteste, soma de {n_seeds} semente(s)", fontsize=8)
    fig.tight_layout()
    show_and_save(fig, path, cfg)


def plot_summary(ci_tbl_raw: pd.DataFrame, path: Path, cfg: Config, metric: str = "qwk") -> None:
    d = ci_tbl_raw.sort_values(f"{metric}_pt")
    labels = [" | ".join(str(r[c]) for c in KEY_COLS if not (c == "scale" and r[c] == 1.0))
              for _, r in d.iterrows()]
    fig, ax = plt.subplots(figsize=(8, 0.32 * len(d) + 1.5))
    err = np.clip(np.nan_to_num(np.vstack([d[f"{metric}_pt"] - d[f"{metric}_lo"],
                                           d[f"{metric}_hi"] - d[f"{metric}_pt"]])), 0, None)
    colors = [plt.cm.tab10(list(MODELS).index(b) if b in MODELS else 9) for b in d.backbone]
    ax.barh(range(len(d)), d[f"{metric}_pt"], xerr=err, color=colors, alpha=0.8, capsize=2)
    ax.set_yticks(range(len(d)), labels, fontsize=7)
    ax.set_xlabel(f"{metric.upper()} de teste (média entre sementes, IC {cfg.ci_level:.0%})")
    ax.grid(axis="x", alpha=0.3)
    fig.tight_layout()
    show_and_save(fig, path, cfg)


# --- Tabelas consolidadas ---
def _pm(g, col):
    m, s = g[col].mean(), g[col].std(ddof=1)
    return m.map(lambda v: "n/a" if pd.isna(v) else f"{v:.4f}") + " ± " + \
        s.map(lambda v: "n/a" if pd.isna(v) else f"{v:.4f}")


def summary_tables(df: pd.DataFrame, cfg: Config):
    """Tabela A: média ± DP entre sementes com metadados. Tabela B: média + IC por bootstrap."""
    g = df.groupby(KEY_COLS, sort=False, dropna=False)
    meta = g.agg(n_sementes=("seed", "nunique"), batch_size=("batch_size", "median"),
                 accum_steps=("accum_steps", "median"), batch_efetivo=("effective_batch", "median"),
                 melhor_epoca=("best_epoch", "mean"), epocas_rodadas=("epochs_run", "mean"),
                 tempo_por_epoca_s=("time_per_epoch_s", "mean"), tempo_ate_melhor_s=("time_to_best_s", "mean"),
                 tempo_total_s=("train_time_s", "mean"), vram_pico_gb=("peak_vram_gb", "max"),
                 C_mediano=("C", "median")).round(3)
    tbl_a = pd.concat([meta, pd.DataFrame({"QWK": _pm(g, "test_qwk"), "F1 macro": _pm(g, "test_f1_macro"),
                                           "Bal. acc": _pm(g, "test_bacc"), "Acc": _pm(g, "test_acc"),
                                           "AUROC": _pm(g, "test_auroc"), "MOD": _pm(g, "test_mod"),
                                           "QWK val": _pm(g, "val_qwk")})], axis=1).reset_index()
    tbl_a = tbl_a.rename(columns=PT_NAMES)

    ci_rows, raw_rows = [], []
    for key, gk in g:
        tbs = {}
        for s in sorted(gk.seed.unique()):
            p = pd.read_csv(PRED_DIR / f"{key_tag(key)}_seed{int(s)}.csv")
            tbs[int(s)] = (p.idx.to_numpy(), p.y_true.to_numpy(), p[[f"p_{c}" for c in CLASS_NAMES]].to_numpy(),
                           p.y_pred.to_numpy())
        ci = holdout_bootstrap_ci(tbs, GROUPS, cfg.n_boot, cfg.ci_level)
        base = dict(zip(KEY_COLS, key))
        ci_rows.append({**{PT_NAMES[k]: v for k, v in base.items()}, "n_sementes": len(tbs),
                        "batch_size": gk.batch_size.median(), "batch_efetivo": gk.effective_batch.median(),
                        "melhor_epoca": round(gk.best_epoch.mean(), 1) if gk.best_epoch.notna().any() else np.nan,
                        **{k: f"{v[0]:.4f} [{v[1]:.4f}, {v[2]:.4f}]" for k, v in ci.items()}})
        raw_rows.append({**base, **{f"{k}_{s}": v[i] for k, v in ci.items()
                                    for i, s in enumerate(("pt", "lo", "hi"))}})
    return tbl_a, pd.DataFrame(ci_rows), pd.DataFrame(raw_rows)


def consolidate(cfg: Config, df: pd.DataFrame, touched: List[tuple]):
    """Tabelas (display + CSV) de TODOS os cenários do run e gráficos dos cenários desta chamada."""
    df = df.copy()
    df["scale"] = df.scale.astype(float).round(6)
    for c in ("batch_size", "accum_steps", "effective_batch", "best_epoch", "epochs_run", "train_time_s",
              "time_to_best_s", "time_per_epoch_s", "peak_vram_gb", "oom_retries", "C", "lbfgs_iter"):
        df[c] = pd.to_numeric(df[c], errors="coerce")
    touched_set = set(touched)
    df_t = df[[norm_key(tuple(r)) in touched_set for r in df[KEY_COLS].itertuples(index=False)]]

    cols = KEY_COLS + ["seed", "batch_role", "batch_size", "accum_steps", "effective_batch", "best_epoch",
                       "epochs_run", "C", "time_per_epoch_s", "time_to_best_s", "train_time_s", "peak_vram_gb",
                       "oom_retries"] + [f"test_{k}" for k in METRIC_KEYS] + ["val_qwk"]
    save_table(df_t[cols].rename(columns={**PT_NAMES, "seed": "Semente", "best_epoch": "Época (melhor)",
                                          "batch_size": "Batch size"}).round(4),
               TABLES_DIR / "execucoes_desta_chamada.csv", "Execuções (cenários desta chamada)")
    tbl_a, tbl_b, raw = summary_tables(df, cfg)
    save_table(tbl_a, TABLES_DIR / "resumo_media_dp.csv",
               "Teste: média ± DP entre sementes (com metadados de batch, épocas e tempo)")
    save_table(tbl_b, TABLES_DIR / "resumo_ic_bootstrap.csv",
               f"Teste: média entre sementes + IC {cfg.ci_level:.0%} (bootstrap de {cfg.n_boot} por imagem)")
    raw.to_csv(TABLES_DIR / "resumo_ic_numerico.csv", index=False)

    for key in touched:
        gk = df[[norm_key(tuple(r)) == key for r in df[KEY_COLS].itertuples(index=False)]]
        if gk.empty:
            continue
        tag, seeds = key_tag(key), sorted(int(s) for s in gk.seed.unique())
        cm = np.zeros((NUM_CLASSES, NUM_CLASSES), dtype=np.int64)
        for s in seeds:
            p = pd.read_csv(PRED_DIR / f"{tag}_seed{s}.csv")
            cm += confusion_matrix(p.y_true, p.y_pred, labels=list(range(NUM_CLASSES)))
        plot_confusion(key, cm, len(seeds), PLOTS_DIR / "confusion_matrices" / f"CM_{tag}.png", cfg)
        hist = {s: pd.read_csv(LOGS_DIR / f"{tag}_seed{s}_history.csv") for s in seeds}
        best = dict(zip(gk.seed.astype(int), gk.best_epoch.astype(int)))
        plot_learning_curves(key, hist, best, PLOTS_DIR / "curvas" / f"aprendizado_{tag}.png", cfg)

    plot_summary(raw, PLOTS_DIR / "resumo_qwk_ic.png", cfg)

    (OUT / "relatorio_comparativo.md").write_text(
        "# Relatório comparativo — fine-tuning completo\n\n"
        f"- Split holdout {cfg.train_frac:.0%}/{cfg.val_frac:.0%}/{cfg.test_frac:.0%} via train_test_split, "
        f"estratificado, {'por imagem de origem' if cfg.split_by_group else 'por ROI'}, repetido por semente; "
        "feito antes do aumento de dados.\n"
        "- Aumento de dados: RCAug em tempo real, só no treino.\n"
        "- Multi-crop: janelas no tamanho nativo do backbone cobrindo toda a ROI; mean pooling das features.\n"
        f"- Fine-tuning completo: AdamW, LR {cfg.lr:g} ({cfg.lr_schedule}), weight decay {cfg.weight_decay:g}, "
        f"{cfg.epochs} épocas, lote {cfg.batch_size} ROIs, loss {cfg.loss}; época escolhida pelo QWK de validação.\n"
        f"- IC {cfg.ci_level:.0%}: bootstrap percentil ({cfg.n_boot} réplicas) reamostrando imagens "
        "de origem (estratificado por classe) dentro do teste de cada semente.\n\n"
        "## Média ± DP entre sementes\n\n" + tbl_a.to_markdown(index=False) +
        "\n\n## Média entre sementes com IC\n\n" + tbl_b.to_markdown(index=False) + "\n")
    print(f"\n[ok] artefatos salvos em {OUT}")
    return {"execucoes": df, "resumo": tbl_a, "ic": tbl_b, "out_dir": OUT}


# ------------------------------- EXECUÇÃO -------------------------------
RESULTS = run_all(CFG)
```
